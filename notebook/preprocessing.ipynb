```python
dt = pd.read_csv("../core/flights.csv")
```

```python
x = ['price']

x = dt[x]

y = dt.drop(columns=x)
```

```python
y = y.drop(columns=['Unnamed: 0'])

# display(y)
```

```python
y['stops'] = y['stops'].map({
    'zero': 0,
    'one': 1,
    'two_or_more': 2
})

# display(y)
```

```python
y['class'] = y['class'].map({
    'Economy' : 0,
    'Business': 1
})

# display(y)
```
